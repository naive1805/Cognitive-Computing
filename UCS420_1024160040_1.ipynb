```python
n = int(input("Enter a number limit: "))

if n < 2:
    print("Your limit was low")
else:
    li = []

    for i in range(2, n + 1):
        for j in range(2, int(i**0.5) + 1):
            if i % j == 0:
                break
        else:
            li.append(i)

    print(li)
    print(sum(li))
```

```python
x = int(input("Enter a number limit: "))

def Sum_of_PN(n):
    if n < 2:
        print("Your limit was low")
        return 0

    li = []

    for i in range(2, n + 1):
        is_prime = True
        for j in range(2, int(i**0.5) + 1):
            if i % j == 0:
                is_prime = False
                break

        if is_prime:
            li.append(i)

    print(li)
    return sum(li)

print(Sum_of_PN(x))
```
