Basic information

```python
print('Shape:', df.shape)
print('Dimensions:', df.ndim)
print('Total elements:', df.size)
print('Data types:')
print(df.dtypes)
print('Describe:')
print(df.describe())
```

statistical information


```python
numeric_columns = ['Study_Hours', 'Sleep_Hours', 'Stress_Level', 'Activity_Level', 'Cognitive_Score', 'Attendance', 'Projects_Completed', 'Satisfaction']

print('Mean values:')
print(df[numeric_columns].mean())

print('\nMedian values:')
print(df[numeric_columns].median())

print('\nMaximum values:')
print(df[numeric_columns].max())

print('\nMinimum values:')
print(df[numeric_columns].min())
```

indexing and slicing

```python
print('Selected rows using loc:')
print(df.loc[[0, 4, 9, 14]])

print('Rows 3 to 7 using loc:')
print(df.loc[3:7])

print('Rows 5 to 9 and selected columns using iloc:')
print(df.iloc[4:9, 2:6])

print('Selected columns using iloc:')
print(df.iloc[:, 1:5])
```

unique values and their frequencies

```python
print('Unique departments:')
print(df['Department'].unique())

print('Number of unique departments:', df['Department'].nunique())

print('Department frequencies:')
print(df['Department'].value_counts())

print('Unique Satisfaction scores:')
print(df['Satisfaction'].unique())
```

idmax and idmin

```python
sorted_by_score = df.sort_values(by='Cognitive_Score', ascending=False)
print('Students sorted by Cognitive_Score (highest to lowest):')
print(sorted_by_score[['Student_ID', 'Cognitive_Score', 'Study_Hours', 'Stress_Level']])

max_score_index = df['Cognitive_Score'].idxmax()
min_score_index = df['Cognitive_Score'].idxmin()

print('Student with maximum Cognitive_Score:')
print(df.loc[max_score_index])

print('Student with minimum Cognitive_Score:')
print(df.loc[min_score_index])
```

derving a value from function

```python
def categorize(score):
    if score >= 85:
        return 'Excellent'
    elif score >= 75:
        return 'Good'
    else:
        return 'Average'

df['Cognitive_Category'] = df['Cognitive_Score'].apply(categorize)

print(df[['Student_ID', 'Cognitive_Score', 'Cognitive_Category']])

print('\nCategory counts:')
print(df['Cognitive_Category'].value_counts())
```

groupby()

```python
department_summary = df.groupby('Department').agg(
    Students=('Student_ID', 'count'),
    Average_Cognitive_Score=('Cognitive_Score', 'mean'),
    Average_Study_Hours=('Study_Hours', 'mean'),
    Average_Stress=('Stress_Level', 'mean'),
    Average_Attendance=('Attendance', 'mean')
)

print(department_summary)

print('Department with highest average Cognitive_Score:')
print(department_summary['Average_Cognitive_Score'].idxmax())

print('Highest average Cognitive_Score:')
print(department_summary['Average_Cognitive_Score'].max())
```
