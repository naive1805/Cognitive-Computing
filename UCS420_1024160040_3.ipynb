```python
import pandas as pd
from google.colab import files
uploaded = files.upload()
df = pd.read_csv("employees.csv")

# Basic information
print("Shape:")
print(df.shape)

print("\nInfo:")
df.info()

print("\nDescriptive Statistics:")
print(df.describe())

# Statistical operations
print("\nMean Salary:")
print(df["Salary"].mean())

print("\nTotal Bonus:")
print(df["Bonus"].sum())

print("\nMinimum Age:")
print(df["Age"].min())

print("\nMaximum Rating:")
print(df["Rating"].max())

# Sort employees by Salary (highest to lowest)
print("\nEmployees sorted by Salary:")
print(df.sort_values(by="Salary", ascending=False))


# Function to categorize performance
def categorize(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"


# Add Performance Category column
df["Performance_Category"] = df["Rating"].apply(categorize)

print("\nPerformance Categories:")
print(df[["Name", "Rating", "Performance_Category"]])


# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())


# Rename Employee_ID to ID
df = df.rename(columns={"Employee_ID": "ID"})

print("\nColumn Names:")
print(df.columns)


# Employees with more than 5 years of experience
exp_more_than_5 = df[df["Years_of_Experience"] > 5]

print("\nEmployees with more than 5 years of experience:")
print(exp_more_than_5)


# Employees from IT department
it_dept = df[df["Department"] == "IT"]

print("\nEmployees in IT department:")
print(it_dept)


# Calculate 10% tax on salary
df["Tax"] = df["Salary"] * 0.10

print("\nSalary and Tax:")
print(df[["Name", "Salary", "Tax"]])


# Save modified DataFrame
df.to_csv("employees_modified.csv", index=False)

print("\nFinal DataFrame:")
print(df.head())
```
