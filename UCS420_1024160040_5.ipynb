Q.2 Questions on Basic NumPy Array: a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5]) b) Find the most frequent value and their indice(s) in the following arrays: i. x = np.array([1,2,3,4,5,1,2,1,1,1]) ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
arr=np.array([1, 2, 3, 6, 4, 5])

rev_arr=arr[::-1]
print("Reversed Array: ", rev_arr)

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
values, counts = np.unique(x, return_counts=True)

most_freq = values[np.argmax(counts)]
print("Most frequent value:", most_freq)
print("Indices:", np.where(x == most_freq)[0])

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
values, counts = np.unique(y, return_counts=True)

most_freq = values[np.argmax(counts)]
print("Most frequent value:", most_freq)
print("Indices:", np.where(y == most_freq)[0])
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.

```python
import numpy as np

ucs420_Namrata = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Namrata)

print("Mean:", np.mean(ucs420_Namrata))

print("Median:", np.median(ucs420_Namrata))

print("Maximum:", np.max(ucs420_Namrata))

print("Minimum:", np.min(ucs420_Namrata))

print("Unique elements:", np.unique(ucs420_Namrata))

reshaped_ucs420_Namrata = ucs420_Namrata.reshape(4, 3)

print("Reshaped Array:")
print(reshaped_ucs420_Namrata)

resized_ucs420_Namrata = np.resize(ucs420_Namrata, (2, 3))

print("Resized Array:")
print(resized_ucs420_Namrata)
```
