3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate). i. ii. Generate a list of 100 random numbers between 100 and 900 (inclusive). Count and print all odd numbers in the list. iii. Count and print all even numbers in the list. iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension. v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
#3
#(i)
import random
from collections import Counter

Roll_Number = 1024160040
random.seed(Roll_Number)

numbers = [random.randint(100, 900) for _ in range(100)]
print("List of 100 random numbers:")
print(numbers)

#(ii)
odd_numbers = [num for num in numbers if num % 2 != 0]
print("\nCount of odd numbers:", len(odd_numbers))

#(iii)
even_numbers = [num for num in numbers if num % 2 == 0]
print("Count of even numbers:", len(even_numbers))

#(iv)
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [num for num in numbers if is_prime(num)]
print("Count of prime numbers:", len(prime_numbers))
print("List of prime numbers found:", prime_numbers)

#(v)
frequency = Counter(numbers)
most_common_number, count = frequency.most_common(1)[0]
print(f"\nMost frequent number: {most_common_number}, occurs {count} times")
```

4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for
each digit} and set B = {digit × 9 for each digit}. Print both sets.
vi. Find the union of A and B — all unique values across both sets.
vii. Find the intersection of A and B — values common to both.
viii. Find A − B and B − A separately (not combined) using difference(), and explain in one line how this
differs from symmetric_difference().
ix. Find the symmetric difference of A and B.
x. Check whether A is a subset of B, and whether B is a superset of A, using issubset() and issuperset().
xi. Ask the user to input a value X. Remove X from set A if present using discard() (not remove()) — explain
in one line why discard() is safer than remove() when you're not sure the value exists.


```python
#4
Roll_Number_8digits = 10241600

digits = [int(d) for d in str(Roll_Number_8digits)]
print("Digits:", digits)

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("\nSet A (digit x 7):", A)
print("Set B (digit x 9):", B)

#(vi) Union
union_AB = A.union(B)
print("\nUnion of A and B:", union_AB)

#(vii) Intersection
intersection_AB = A.intersection(B)
print("Intersection of A and B:", intersection_AB)

#(viii) Difference
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print("\nA - B:", A_minus_B)
print("B - A:", B_minus_A)
# difference() gives elements unique to ONE specific set (one-directional),
# while symmetric_difference() gives elements that are in either set but not both (combines A-B and B-A together).

#(ix) Symmetric difference
sym_diff_AB = A.symmetric_difference(B)
print("\nSymmetric difference of A and B:", sym_diff_AB)

#(x) Subset and Superset check
print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

#(xi) Discard user input value from A
X = int(input("Enter a value X to remove from set A: "))
A.discard(X)
print("\nSet A after discarding X:", A)
# discard() is safer than remove() because discard() does nothing if the value is not found,
# whereas remove() raises a KeyError if the value doesn't exist in the set.
```

my_dict = {

"roll_no": "<your roll number>",

"age": <your age>,
"city": "<your home city>"
}
i. Rename the key “city” to “location” without changing its value (do not just recreate the dictionary by
hand — use pop() or dictionary comprehension so the code would work for any dictionary).
ii. Add a new key “cgpa” with your CGPA as its value.
iii. Update the value of “age” by increasing it by 1.
iv. Delete the “branch” key using two different methods in two separate copies of the dictionary: once
with pop() and once with the del keyword. Explain in one line how pop() differs from del (hint: what
does each return?).
v. Iterate over your dictionary using .items() and print each key-value pair as “key → value”.
vi. Before accessing a key called “email” (which doesn't exist in your dictionary), check for its existence
using the in keyword and print a safe fallback message instead of letting the program crash.
vii. Create a second dictionary friend_dict with the same 5 original keys but a friend's (fictional) details.
Merge it with your own using the {**dict1, **dict2} syntax, and explain in one line which values “win”
when both dictionaries share a key.
viii. Use dictionary comprehension to create a new dictionary containing only the key-value pairs from your
original dictionary where the value is a string.

```python
#5
my_dict = {
    "name": "Namrata",
    "roll_no": "1024160040",
    "branch": "Computer Science",
    "age": 20,
    "city": "Amritsar"
}
print("Original dictionary:", my_dict)

#(i) Rename "city" to "location" using pop() - works for any dictionary
my_dict["location"] = my_dict.pop("city")
print("\nAfter renaming 'city' to 'location':", my_dict)

#(ii) Add a new key "cgpa"
my_dict["cgpa"] = 8.5
print("\nAfter adding 'cgpa':", my_dict)

#(iii) Update "age" by increasing it by 1
my_dict["age"] += 1
print("\nAfter updating 'age':", my_dict)

#(iv) Delete "branch" key using pop() and del in two separate copies
dict_copy1 = my_dict.copy()
dict_copy2 = my_dict.copy()

removed_value = dict_copy1.pop("branch")
print("\nUsing pop(): removed value =", removed_value)
print("Dictionary after pop():", dict_copy1)

del dict_copy2["branch"]
print("\nDictionary after del:", dict_copy2)
# pop() returns the value of the removed key (useful if you need it),
# while del simply removes the key-value pair and returns nothing.

#(v) Iterate using .items()
print("\nIterating over dictionary:")
for key, value in my_dict.items():
    print(f"{key} -> {value}")

#(vi) Safely check for "email" key using in keyword
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nKey 'email' does not exist in the dictionary.")

#(vii) Create friend_dict with same 5 original keys, merge using {**dict1, **dict2}
friend_dict = {
    "name": "Ishwinder",
    "roll_no": "1024160043",
    "branch": "Computer Science",
    "age": 21,
    "city": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}
print("\nMerged dictionary:", merged_dict)
# When both dictionaries share a key, the values from the dictionary that comes
# LAST (friend_dict here, since it's second in {**my_dict, **friend_dict}) win/overwrite the earlier one.

#(viii) Dictionary comprehension - only key-value pairs where value is a string
string_value_dict = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print("\nDictionary with only string values:", string_value_dict)
```
